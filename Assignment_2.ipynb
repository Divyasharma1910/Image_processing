```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
img = cv2.imread(next(iter(uploaded)), cv2.IMREAD_COLOR)

if img is None:
    raise ValueError("Image could not be loaded.")

print("Image loaded successfully:", img.shape)
```

```python
B, G, R = cv2.split(img)

print("Blue, Green and Red channels extracted successfully.")
```

```python
gray_mean = ((R.astype(float) + G.astype(float) + B.astype(float)) / 3).astype(np.uint8)
```

```python
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
plt.imshow(gray_mean, cmap='gray')
plt.axis('off')
plt.show()
```

```python
wr = float(input("Enter weight for R: "))
wg = float(input("Enter weight for G: "))
wb = float(input("Enter weight for B: "))
```

```python
if abs(wr + wg + wb - 1) > 0.0001:
    print("Invalid weights. Sum must be equal to 1.")
```

```python
gray_weighted = (wr * R.astype(float) + wg * G.astype(float) + wb * B.astype(float)).astype(np.uint8)
```

```python
plt.imshow(gray_weighted, cmap='gray')
plt.axis('off')
plt.show()
```
